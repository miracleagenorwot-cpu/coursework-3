## q4: 5 marks
Write efficiency(out_w, in_w) returning the percentage, and returning None when in_w is zero or negative.

```python
# Your answer here
def efficiency(out_w, in_w):
    """Calculates efficiency percentage, returning None for zero/negative inputs."""
    if in_w <= 0:
        return None
    return 100 * out_w / in_w
```

## q5: 5 marks
READINGS is given in the notebook. Using your own percent_error against a nominal of 100, set worst_reading to the reading with the largest error by size, and worst_error to that reading's signed percent error.

You may compare these six readings directly; a loop is not required this week.

```python
READINGS = [98.5, 103.2, 99.0, 111.4, 94.8, 100.6]
# Assuming READINGS is pre-loaded in your notebook environment
# 1. Generate errors using your percent_error function
errors = [percent_error(r, 100) for r in READINGS]

# 2. Find the absolute magnitudes to evaluate size
sizes = [abs(e) for e in errors]

# 3. Locate the index of the maximum error magnitude
position = sizes.index(max(sizes))

# 4. Extract the worst reading and its signed error
worst_reading = READINGS[position]
worst_error = errors[position]

print(worst_reading, worst_error)
```
